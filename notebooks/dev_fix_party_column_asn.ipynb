```python
import os
import pandas as pd

from project_huishoudboekje.config import GeneralSettings as GenSet

from project_huishoudboekje.read_asn import ReadAsn
```

```python
list_files = next(os.walk(GenSet.project_path / f'data/ASN Bank'), (None, None, []))[2]
```
